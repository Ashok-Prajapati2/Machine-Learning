```python
import matplotlib.pyplot as plt
import seaborn as sns 
import pandas as pd
import numpy as np
%matplotlib inline
```

```python
# Load dataset
from sklearn.datasets import load_breast_cancer
```

```python
df1 = load_breast_cancer()
df = df1
df.keys()
```

```python
df.data.shape
```

```python
print(df.DESCR)
```

```python
df = pd.DataFrame(df.data, columns=df.feature_names)
```

```python
df.head()
```

```python
sns.pairplot(df, vars=['mean radius', 'mean texture', 'mean area', 'mean perimeter', 'mean smoothness'] )
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
scaler.fit(df)
```

```python
scaler_data = scaler.transform(df)
```

```python
scaled_df = pd.DataFrame(scaler_data, columns=df.columns)
```

```python
scaled_df.head()
```

```python
## apply pca to scaled data(principal component analysis)
```

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
```

```python
pca_data = pca.fit_transform(scaler_data)
```

```python
pca_df = pd.DataFrame(pca_data, columns=['PC1', 'PC2'])
```

```python
pca_df.head()
```

```python
sns.scatterplot(x='PC1', y='PC2', data=pca_df)
```

```python
pca.explained_variance_
```

```python
pca.explained_variance_ratio_
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(pca_df['PC1'], pca_df['PC2'], c=df1['target'], cmap='viridis')
plt.xlabel('First Principal Component')
plt.ylabel('Second Principal Component')
plt.show()
```
