```python
import random
import pandas as pd

# Step 1: Define weights for two clusters
weights_cluster_1 = [i for i in range(1, 25)]  # First cluster (smaller values)
weights_cluster_2 = [i for i in range(1, 25)]  # Second cluster (larger values)

# Define populations for two clusters
population_cluster_1 = list(range(1, 25))  # Cluster 1 with values from 1 to 24
population_cluster_2 = list(range(50, 74))  # Cluster 2 with values from 50 to 74

# Step 2: Sample from the two clusters
df1 = random.choices(population_cluster_1, k=50, weights=weights_cluster_1)  # 50 points from cluster 1
df2 = random.choices(population_cluster_2, k=50, weights=weights_cluster_2)  # 50 points from cluster 2

# Combine the two sets of samples
df1.extend(df2)

# Step 3: Create the DataFrame
df = pd.DataFrame({'A': df1, 'B': df1})

# Show the first few rows of the dataset
print(df.head())

# Optionally, plot the dataset to visualize the clusters
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
plt.scatter(df['A'], df['B'], c='blue', label='Clustered Data')
plt.title('2-Cluster Dataset')
plt.xlabel('Feature 1 (A)')
plt.ylabel('Feature 2 (B)')
plt.legend()
plt.show()
```
